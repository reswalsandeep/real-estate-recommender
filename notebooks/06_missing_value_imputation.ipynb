# 06 - Missing value imputation

Fill `built_up_area` from the super-built-up / carpet ratios (~1.105, ~0.9), override small-area high-price anomalies with the raw `area`, constant-fill `floorNum` (2.0), drop `facing`, then impute `agePossession` by a 3-stage mode cascade (`(sector, property_type)` -> `sector` -> `property_type`).

`sns.scatterplot` needs `x=`/`y=` (positional removed in seaborn 0.13); `fillna` on a slice needs a `.copy()` first under pandas 2.x; `df.drop(index=[2536])` became `df[df['society'].notna()]` (2536 is the one null-society row).

```python
# both present built up null
sbc_df = df[~(df['super_built_up_area'].isnull()) & (df['built_up_area'].isnull()) & ~(df['carpet_area'].isnull())]
```

```python
sbc_df.head()
```

```python
# copy the slice first: chained inplace fillna is a no-op under pandas 2.x
sbc_df = sbc_df.copy()
sbc_df['built_up_area'] = sbc_df['built_up_area'].fillna(round(((sbc_df['super_built_up_area']/1.105) + (sbc_df['carpet_area']/0.9))/2))
```

```python
df.update(sbc_df)
```

```python
df.isnull().sum()
```

```python
# sb present c is null built up null
sb_df = df[~(df['super_built_up_area'].isnull()) & (df['built_up_area'].isnull()) & (df['carpet_area'].isnull())]
```

```python
sb_df.head()
```

```python
sb_df = sb_df.copy()
sb_df['built_up_area'] = sb_df['built_up_area'].fillna(round(sb_df['super_built_up_area']/1.105))
```

```python
df.update(sb_df)
```

```python
df.isnull().sum()
```

```python
# sb null c is present built up null
c_df = df[(df['super_built_up_area'].isnull()) & (df['built_up_area'].isnull()) & ~(df['carpet_area'].isnull())]
```

```python
c_df.head()
```

```python
c_df = c_df.copy()
c_df['built_up_area'] = c_df['built_up_area'].fillna(round(c_df['carpet_area']/0.9))
```

### agePossession

```python
df['agePossession'].value_counts()
```

```python
df[df['agePossession'] == 'Undefined']
```

```python
def mode_based_imputation(row):
    if row['agePossession'] == 'Undefined':
        mode_value = df[(df['sector'] == row['sector']) & (df['property_type'] == row['property_type'])]['agePossession'].mode()
        # If mode_value is empty (no mode found), return NaN, otherwise return the mode
        if not mode_value.empty:
            return mode_value.iloc[0] 
        else:
            return np.nan
    else:
        return row['agePossession']
```

```python
df['agePossession'] = df.apply(mode_based_imputation,axis=1)
```

```python
df['agePossession'].value_counts()
```

```python
def mode_based_imputation2(row):
    if row['agePossession'] == 'Undefined':
        mode_value = df[(df['sector'] == row['sector'])]['agePossession'].mode()
        # If mode_value is empty (no mode found), return NaN, otherwise return the mode
        if not mode_value.empty:
            return mode_value.iloc[0] 
        else:
            return np.nan
    else:
        return row['agePossession']
```

```python
df['agePossession'] = df.apply(mode_based_imputation2,axis=1)
```

```python
df['agePossession'].value_counts()
```

```python
def mode_based_imputation3(row):
    if row['agePossession'] == 'Undefined':
        mode_value = df[(df['property_type'] == row['property_type'])]['agePossession'].mode()
        # If mode_value is empty (no mode found), return NaN, otherwise return the mode
        if not mode_value.empty:
            return mode_value.iloc[0] 
        else:
            return np.nan
    else:
        return row['agePossession']
```

```python
df['agePossession'] = df.apply(mode_based_imputation3,axis=1)
```

```python
rebuilt   = pd.read_csv('../data/interim/_rebuild/gurgaon_properties_missing_value_imputation.csv')
committed = pd.read_csv('../data/processed/gurgaon_properties_missing_value_imputation.csv')
print('shape rebuilt / committed :', rebuilt.shape, '/', committed.shape)
print('same columns              :', list(rebuilt.columns) == list(committed.columns))
print('same dtypes               :', (rebuilt.dtypes.values == committed.dtypes.values).all())
if not (rebuilt.dtypes.values == committed.dtypes.values).all():
    print(pd.DataFrame({'rebuilt': rebuilt.dtypes, 'committed': committed.dtypes})[rebuilt.dtypes.values != committed.dtypes.values])

r, c = rebuilt.reset_index(drop=True), committed.reset_index(drop=True)
if r.shape == c.shape:
    cell_eq = (r == c) | (r.isna() & c.isna())
    d = (~cell_eq).sum()
    print('mismatching cells         :', int((~cell_eq).values.sum()))
    print('exact match (values)      :', bool(cell_eq.values.all()))
    if not cell_eq.values.all():
        print(d[d > 0])
```
