# 05 - Outlier treatment

Drop duplicates and extreme rows, rescale square-yard areas, recompute `price_per_sqft`. 14 `sns.distplot` calls became `sns.histplot` (removed in seaborn 0.14); `df.update` on a slice needs a `.copy()` first under pandas 2.x.

The original notebook has no `to_csv` and its cells leave `df` at 3588 rows / 23 columns. The committed `outlier_treated.csv` (3555 / 24) also has an `area_room_ratio` column, 33 dropped rows and 70 `bedRoom` fixes with no code anywhere - those are applied at the end from constant tables (positional indices into `cleaned_v2.csv`), recovered from the input/output diff. The 33 drops are a subset of the `area/bedRoom < 183` rows; the 70 `bedRoom` fixes aren't a formula or a join.

```python
# Calculate the IQR for the 'price' column
Q1 = df['price'].quantile(0.25)
Q3 = df['price'].quantile(0.75)
IQR = Q3 - Q1

# Define bounds for outliers
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Identify outliers
outliers = df[(df['price'] < lower_bound) | (df['price'] > upper_bound)]

# Displaying the number of outliers and some statistics
num_outliers = outliers.shape[0]
outliers_price_stats = outliers['price'].describe()

num_outliers, outliers_price_stats
```

```python
# Calculate the IQR for the 'price' column
Q1 = df['price_per_sqft'].quantile(0.25)
Q3 = df['price_per_sqft'].quantile(0.75)
IQR = Q3 - Q1

# Define bounds for outliers
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Identify outliers
outliers_sqft = df[(df['price_per_sqft'] < lower_bound) | (df['price_per_sqft'] > upper_bound)]

# Displaying the number of outliers and some statistics
num_outliers = outliers_sqft.shape[0]
outliers_sqft_stats = outliers_sqft['price_per_sqft'].describe()

num_outliers, outliers_sqft_stats
```

```python
outliers_sqft = outliers_sqft.copy()  # was implicitly a slice of df; mutating it in place is a no-op under pandas 2.x CoW
outliers_sqft['area'] = outliers_sqft['area'].apply(lambda x:x*9 if x<1000 else x)
```

```python
outliers_sqft['price_per_sqft'] = round((outliers_sqft['price']*10000000)/outliers_sqft['area'])
```

```python
outliers_sqft['price_per_sqft'].describe()
```

```python
df.update(outliers_sqft)

# pandas < 2.0 DataFrame.update upcast every column present in `other` to float64;
# pandas >= 2.0 preserves dtype. The committed outlier_treated.csv was built on the
# old behaviour and carries these integer columns as float, so reproduce that.
for _col in ['bedRoom','bathroom','study room','servant room','store room',
             'pooja room','others','furnishing_type','luxury_score']:
    df[_col] = df[_col].astype(float)
```

## Reconstruction gap

The committed `outlier_treated.csv` has an `area_room_ratio` column, 33 dropped rows and 70 `bedRoom` fixes that no cell above produces. They come from the `cleaned_v2 -> outlier_treated` diff and are applied below as constants, keyed by positional row index into `cleaned_v2.csv`.

```python
# state after the visible cells
df_visible = df.copy()
committed = pd.read_csv('../data/interim/gurgaon_properties_outlier_treated.csv')
print('after the 81 cells :', df_visible.shape)
print('committed output   :', committed.shape)
print('columns only in committed :', [c for c in committed.columns if c not in df_visible.columns])
print('row-count gap              :', df_visible.shape[0] - committed.shape[0], '(rows the notebook does not drop)')
```

```python
# 33 low area/bedRoom rows dropped by hand, recovered from the diff (a subset
# of the area/bedRoom < 183 rows; positional indices into cleaned_v2.csv)
_RECONSTRUCTED_ROW_DROP = [
    37, 93, 229, 247, 387, 751, 753, 1106, 1206, 1429, 1562, 1580, 1696, 1737,
    1747, 1773, 1936, 1939, 1953, 1997, 2047, 2300, 2360, 2721, 2784, 2806,
    3148, 3246, 3268, 3306, 3329, 3633, 3774,
]
print(len(_RECONSTRUCTED_ROW_DROP), 'rows to drop')
```

```python
# 70 house rows with bedRoom corrected down (4-10 -> 1-4); not a formula or a
# join to houses.csv, recovered from the diff
_RECONSTRUCTED_HOUSE_BEDROOM = {
    9: 3, 15: 2, 48: 3, 74: 2, 99: 3, 140: 2, 186: 2, 255: 2, 293: 2, 343: 3,
    393: 3, 530: 2, 540: 3, 565: 2, 668: 3, 837: 4, 848: 2, 852: 3, 880: 3,
    886: 1, 935: 2, 1033: 2, 1049: 3, 1087: 3, 1167: 2, 1187: 3, 1213: 2,
    1224: 2, 1384: 2, 1407: 3, 1480: 4, 1500: 2, 1509: 2, 1524: 2, 1532: 1,
    1537: 2, 1627: 2, 1798: 3, 1851: 1, 1951: 2, 2004: 2, 2063: 3, 2173: 2,
    2181: 3, 2244: 2, 2261: 4, 2266: 4, 2277: 1, 2313: 2, 2333: 3, 2343: 2,
    2386: 2, 2461: 4, 2660: 2, 2722: 2, 2984: 2, 3045: 3, 3101: 3, 3198: 2,
    3252: 3, 3254: 2, 3282: 2, 3309: 2, 3328: 2, 3347: 2, 3364: 1, 3546: 2,
    3686: 2, 3713: 2, 3738: 2,
}
print(len(_RECONSTRUCTED_HOUSE_BEDROOM), 'house rows to correct')
```

```python
# apply the drops + bedRoom fixes, add area_room_ratio, reset index
df = df_visible.drop(index=[i for i in _RECONSTRUCTED_ROW_DROP if i in df_visible.index])

_fixed = 0
for _idx, _val in _RECONSTRUCTED_HOUSE_BEDROOM.items():
    if _idx in df.index:
        df.loc[_idx, 'bedRoom'] = _val
        _fixed += 1
print('bedRoom corrections applied:', _fixed, '/', len(_RECONSTRUCTED_HOUSE_BEDROOM))

df['area_room_ratio'] = df['area'] / df['bedRoom']   # no cell in the original creates this
df = df.reset_index(drop=True)

import os
os.makedirs('../data/interim/_rebuild', exist_ok=True)
df.to_csv('../data/interim/_rebuild/gurgaon_properties_outlier_treated.csv', index=False)
print('wrote sidecar:', df.shape)
```

### Check against the committed file

```python
rebuilt   = pd.read_csv('../data/interim/_rebuild/gurgaon_properties_outlier_treated.csv')
committed = pd.read_csv('../data/interim/gurgaon_properties_outlier_treated.csv')
print('shape rebuilt / committed :', rebuilt.shape, '/', committed.shape)
print('same columns              :', list(rebuilt.columns) == list(committed.columns))
print('same dtypes               :', (rebuilt.dtypes.values == committed.dtypes.values).all())
if not (rebuilt.dtypes.values == committed.dtypes.values).all():
    print(pd.DataFrame({'rebuilt': rebuilt.dtypes, 'committed': committed.dtypes})[rebuilt.dtypes.values != committed.dtypes.values])

r, c = rebuilt.reset_index(drop=True), committed.reset_index(drop=True)
if r.shape == c.shape and list(r.columns) == list(c.columns):
    cell_eq = (r == c) | (r.isna() & c.isna())
    print('total cells               :', r.size)
    print('mismatching cells         :', int((~cell_eq).values.sum()))
    print('exact match (values)      :', bool(cell_eq.values.all()))
    if not cell_eq.values.all():
        d = (~cell_eq).sum(); print(d[d > 0])
```
