# 00 - Data collection

The five files in `data/raw/` are 99acres.com scrapes of Gurgaon-area listings, from CampusX's DSMP real-estate capstone. No scraping code was provided. Nothing is processed here - this notebook just documents the inputs.

```python
from pathlib import Path
import pandas as pd
from IPython.display import display

REPO_ROOT = Path.cwd().parents[0] if Path.cwd().name == "notebooks" else Path.cwd()
RAW = REPO_ROOT / "data" / "raw"

RAW_FILES = [
    "flats.csv",
    "houses.csv",
    "independent-house.csv",
    "residential-land.csv",
    "appartments.csv",
]
```

## The raw files

| file | what it is |
|---|---|
| `flats.csv` | individual flat/apartment listings |
| `houses.csv` | individual independent-house listings |
| `independent-house.csv` | the **same scrape** as `houses.csv` (shared `property_id`s), with 51 extra raw rows |
| `residential-land.csv` | plot / residential-land listings |
| `appartments.csv` | project-level records (per-BHK price bands, amenities, landmark distances) — one row per development, not per listing |

```python
summary = []
for name in RAW_FILES:
    df = pd.read_csv(RAW / name)
    summary.append({"file": name, "rows": df.shape[0], "cols": df.shape[1]})
pd.DataFrame(summary).set_index("file")
```

```python
for name in RAW_FILES:
    df = pd.read_csv(RAW / name)
    print(f"\n=== {name}  {df.shape} ===")
    print("columns:", list(df.columns))
    display(df.head(2))
```

### `houses.csv` and `independent-house.csv` are one scrape

**Finding:** every real
`property_id` in `houses.csv` also appears in `independent-house.csv` and vice
versa — `independent-house.csv` just has more raw rows. Merging them as if
additive would double-count ~1,000 properties. The pipeline uses `houses.csv`.

Counting **non-null** ids (both files have some blank `property_id` rows —
8 and 59 respectively):

```python
h_ids = pd.read_csv(RAW / "houses.csv")["property_id"].dropna()
ih_ids = pd.read_csv(RAW / "independent-house.csv")["property_id"].dropna()
h, ih = set(h_ids), set(ih_ids)
print(f"houses.csv           : {len(h_ids)} non-null property_id rows, {len(h)} distinct")
print(f"independent-house.csv : {len(ih_ids)} non-null property_id rows, {len(ih)} distinct")
print(f"distinct ids in common          : {len(h & ih)}")
print(f"distinct ids only in houses.csv : {len(h - ih)}")
print(f"distinct ids only in indep-house: {len(ih - h)}")
```
